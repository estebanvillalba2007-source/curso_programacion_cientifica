# Ejercicio 1 — Calibrar una lectura
Escriba la función que aplica la recta de calibración a una lectura. Sus parámetros de calibración son opcionales: si no se indican, la función no altera la lectura (pendiente 1.0 e intercepto 0.0).

```python
def calibrar_lectura(valor, pendiente=1.0, intercepto=0.0):
    """Aplica la recta de calibración a una lectura."""
    return valor * pendiente + intercepto


sin_calibrar = calibrar_lectura(10.0)

calibrada = calibrar_lectura(10.0, 1.2, -0.5)

con_intercepto = calibrar_lectura(10.0, intercepto=2.0)

print("Sin calibrar:", sin_calibrar)
print("Calibrada:", calibrada)
print("Con intercepto:", con_intercepto)
```

# Ejercicio 2 — ¿Es utilizable la lectura?
Una lectura es válida cuando no es el centinela -999.0 y está entre 0 y 500.0 inclusive.

```python
lecturas_sbc_a = [
    10.0,
    -999.0,
    25.0,
    40.0,
    -999.0,
    8.0,
    52.0,
    600.0
]


def es_lectura_valida(valor):
    """Determina si una lectura está dentro del rango válido."""
    return valor != -999.0 and 0 <= valor <= 500.0


print(es_lectura_valida(25.0))
print(es_lectura_valida(-999.0))
print(es_lectura_valida(600.0))
print(es_lectura_valida(0.0))
```

# Ejercicio 3 — Promedio de las lecturas útiles
Ahora reutilice la función del Ejercicio 2 dentro de otra función.

```python
def promedio_valido(lecturas):
    """Calcula el promedio de las lecturas válidas."""

    suma = 0.0
    contador = 0

    for lectura in lecturas:

        if es_lectura_valida(lectura):
            suma += lectura
            contador += 1

    if contador == 0:
        return None

    return suma / contador


promedio_sbc_a = promedio_valido(lecturas_sbc_a)

print("Promedio SBC-A:", promedio_sbc_a)

print(
    "Promedio sin lecturas válidas:",
    promedio_valido([-999.0, -999.0])
)
```

# Ejercicio 4 — Calibrar la serie con map, filter y lambda
Con las funciones anteriores, calibre solo las lecturas útiles de SBC-A y cuente cuántas superan la categoría Moderada.

```python
pendiente_sbc_a = 1.2
intercepto_sbc_a = -0.5


lecturas_utiles = list(
    filter(es_lectura_valida, lecturas_sbc_a)
)


lecturas_calibradas = list(
    map(
        lambda lectura: calibrar_lectura(
            lectura,
            pendiente_sbc_a,
            intercepto_sbc_a
        ),
        lecturas_utiles
    )
)


excedencias = len(
    list(
        filter(
            lambda lectura: lectura > 35.4,
            lecturas_calibradas
        )
    )
)


print("Lecturas calibradas:", lecturas_calibradas)
print("Excedencias:", excedencias)
```

# Ejercicio 5 — La clase Lectura
Convierta una lectura suelta en un objeto que sabe convertirse y sabe si es un dato faltante. En este ejercicio la clase no valida: el centinela -999.0 es un valor legítimo que debe poder guardarse.

```python

class Lectura:

    def __init__(
        self,
        valor,
        unidad="ug/m3",
        codigo_sensor="sin asignar"
    ):
        """Inicializa una lectura."""

        self.valor = valor
        self.unidad = unidad
        self.codigo_sensor = codigo_sensor


    def a_microgramos(self):
        """Convierte la lectura a microgramos por metro cúbico."""

        if self.unidad == "mg/m3":
            return self.valor * 1000

        return self.valor


    def es_dato_faltante(self):
        """Determina si la lectura corresponde al dato faltante."""

        return self.valor == -999.0


lectura_campo = Lectura(
    0.0252,
    "mg/m3",
    "SBC-A"
)

lectura_faltante = Lectura(
    -999.0,
    codigo_sensor="SBC-A"
)

lectura_normal = Lectura(30.0)


print(lectura_campo.a_microgramos())

print(lectura_faltante.es_dato_faltante())

print(lectura_campo.es_dato_faltante())

print(
    lectura_normal.unidad,
    lectura_normal.codigo_sensor
)
```

# Ejercicio 6 — La clase Sensor: validación y encapsulamiento
Un sensor con una pendiente de calibración inválida produciría resultados absurdos en silencio. La clase debe impedir que ese objeto exista.

```python
class Sensor:

    def __init__(
        self,
        codigo,
        pendiente=1.0,
        intercepto=0.0
    ):
        """Inicializa un sensor con parámetros de calibración válidos."""

        if codigo == "":
            raise ValueError("El código del sensor no puede estar vacío")

        if pendiente <= 0:
            raise ValueError("La pendiente debe ser mayor que cero")

        self.codigo = codigo
        self._pendiente = pendiente
        self._intercepto = intercepto


    def calibrar(self, valor):
        """Calibra una lectura usando los parámetros del sensor."""

        return calibrar_lectura(
            valor,
            self._pendiente,
            self._intercepto
        )


    def __repr__(self):
        """Devuelve una representación técnica del sensor."""

        return (
            f"Sensor(codigo='{self.codigo}', "
            f"pendiente={self._pendiente}, "
            f"intercepto={self._intercepto})"
        )


    def __str__(self):
        """Devuelve una descripción legible del sensor."""

        return (
            f"Sensor {self.codigo}: "
            f"pendiente={self._pendiente}, "
            f"intercepto={self._intercepto}"
        )


sensor_a = Sensor(
    "SBC-A",
    1.2,
    -0.5
)

sensor_b = Sensor("SBC-B")


print(sensor_a)

print([sensor_a, sensor_b])

print(sensor_a.calibrar(10.0))

print(sensor_b.calibrar(10.0))
```

# Ejercicio 7 — La clase Estacion: objetos que contienen objetos
La estación EST-04 guarda las lecturas de su sensor como objetos Lectura del Ejercicio 5. Las lecturas de la campaña, como pares de valor y unidad, son estas (una llegó en mg/m³):

```python
class Estacion:

    def __init__(self, codigo, sector):
        """Inicializa una estación sin lecturas."""

        self.codigo = codigo
        self.sector = sector
        self._lecturas = []


    def agregar(self, lectura):
        """Agrega una lectura a la estación."""

        self._lecturas.append(lectura)


    def valores(self):
        """Devuelve las lecturas convertidas a microgramos."""

        return [
            lectura.a_microgramos()
            for lectura in self._lecturas
        ]


    def __repr__(self):
        """Devuelve una representación técnica de la estación."""

        return (
            f"Estacion('{self.codigo}', "
            f"'{self.sector}', "
            f"lecturas={len(self._lecturas)})"
        )


datos_campana = [
    (10.0, "ug/m3"),
    (-999.0, "ug/m3"),
    (0.025, "mg/m3"),
    (40.0, "ug/m3"),
    (-999.0, "ug/m3"),
    (8.0, "ug/m3"),
    (52.0, "ug/m3"),
    (600.0, "ug/m3")
]

estacion_oriente = Estacion(
    "EST-04",
    "Oriente"
)

for valor, unidad in datos_campana:

    lectura = Lectura(
        valor,
        unidad,
        "SBC-A"
    )

    estacion_oriente.agregar(lectura)

print(estacion_oriente)
print(estacion_oriente.valores())
```

# Ejercicio 8 — Integración: el informe de la estación
El último ejercicio junta funciones y objetos: una clase cuyos métodos trabajan con funciones auxiliares que usted ya escribió.

```python
def categoria_pm25(concentracion):
    """Clasifica una concentración de PM2.5."""

    if concentracion <= 12.0:
        return "Buena"

    elif concentracion <= 35.4:
        return "Moderada"

    elif concentracion <= 55.4:
        return "Dañina para grupos sensibles"

    else:
        return "Dañina para la salud"


class InformeEstacion:

    def __init__(self, estacion, sensor):
        """Inicializa un informe con estación y sensor."""

        self._estacion = estacion
        self._sensor = sensor


    def promedio_calibrado(self):
        """Calcula el promedio de las lecturas válidas calibradas."""

        valores = self._estacion.valores()

        valores_validos = list(
            filter(
                es_lectura_valida,
                valores
            )
        )

        valores_calibrados = list(
            map(
                self._sensor.calibrar,
                valores_validos
            )
        )

        return promedio_valido(
            valores_calibrados
        )


    def categoria(self):
        """Devuelve la categoría del promedio calibrado."""

        return categoria_pm25(
            self.promedio_calibrado()
        )


    def __str__(self):
        """Devuelve el informe resumido de la estación."""

        return (
            f"{self._estacion.codigo}: "
            f"{round(self.promedio_calibrado(), 1)} ug/m3 - "
            f"{self.categoria()}"
        )


informe = InformeEstacion(
    estacion_oriente,
    sensor_a
)

print(informe)
print(informe.promedio_calibrado())
```
