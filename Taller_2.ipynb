Ejercicio 6 — Recorrer el inventario completo
La tienda tiene ahora un inventario con varios productos:

```python
cantidades_en_stock = [0, 3, 12, 0, 7, 2]

for indice in range(len(cantidades_en_stock)):
    cantidad = cantidades_en_stock[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    print(f"Producto {indice}: {nivel}")
```

Ejercicio 7 — Promedio semanal de temperaturas
La semana de temperaturas quedó registrada así:

```python
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]

suma_temperaturas = 0
dias_calurosos = 0

for indice in range(len(temperaturas_semana)):
    temperatura = temperaturas_semana[indice]

    suma_temperaturas = suma_temperaturas + temperatura

    if temperatura > 20:
        dias_calurosos = dias_calurosos + 1

promedio = suma_temperaturas / len(temperaturas_semana)

print("Promedio semanal:", promedio)
print("Días calurosos:", dias_calurosos)
```

Ejercicio 8 — Comprar tornillos mientras alcance el presupuesto
La ferretería tiene un presupuesto para reponer tornillos:

```python
presupuesto_ferreteria = 3000
precio_tornillo = 250

tornillos_comprados = 0

while presupuesto_ferreteria >= precio_tornillo:
    presupuesto_ferreteria = presupuesto_ferreteria - precio_tornillo
    tornillos_comprados = tornillos_comprados + 1

print("Tornillos comprados:", tornillos_comprados)
print("Presupuesto restante:", presupuesto_ferreteria)
```

Ejercicio 9 — Buscar la primera herramienta cara
La ferretería tiene los siguientes precios de herramientas:

```python
precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

for precio in precios_herramientas:
    if precio > 20000:
        print(f"Primera herramienta cara encontrada: ${precio}")
        break
    else:
        print(f"Sigue revisando: ${precio}")
```

Ejercicio 10 — Revisión completa del inventario de la ferretería
La ferretería tiene el siguiente inventario, con nombre y cantidad de cada herramienta en la misma posición de cada lista

```python
nombres_herramientas = [
    "Martillo",
    "Taladro",
    "Sierra",
    "Destornillador",
    "Nivel",
    "Llave inglesa"
]

cantidades_herramientas = [5, 0, 2, 8, 0, 1]

for indice in range(len(nombres_herramientas)):
    nombre = nombres_herramientas[indice]
    cantidad = cantidades_herramientas[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    if nivel == "Stock suficiente":
        continue

    elif nivel == "Agotado":
        print(f"Alerta: reponer urgentemente {nombre}")
        break

    else:
        print(f"{nombre}: {cantidad} unidades")
```
