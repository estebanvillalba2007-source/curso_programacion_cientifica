Ejercicio 1 — Inventario de la tienda

Una tienda pequeña tiene el siguiente inventario:

Productos y precios (en pesos colombianos):
- Cuaderno: 3500
- Lápiz: 800
- Café: 6200.50
Cantidad a comprar: 3 cuadernos, 5 lápices
La tienda tiene actualmente descuentos activos: no



```python
nombre_producto = "cuaderno"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print(type(nombre_producto))
print(type(precio_cuaderno))
print(type(precio_lapiz))
print(type(precio_cafe))
print(type(hay_descuento))
print(type(cantidad_cuadernos))
print(type(cantidad_lapices))
```

Ejercicio 2 — Total de la compra y descuento
Retome las variables del Ejercicio 1.

```python
# Calcular total de cuadernos

total_cuadernos = precio_cuaderno * cantidad_cuadernos
print(total_cuadernos)

# Calcular total de lápices

total_lapices = precio_lapiz * cantidad_lapices
print(total_lapices)

# Calcular total general

total_compra = total_cuadernos + total_lapices + precio_cafe
print(total_compra)

# Calcular descuento del 10%

descuento = total_compra * 0.10
print(descuento)

# Calcular total con descuento

total_pagar = total_compra - descuento
print(total_pagar)
```

Ejercicio 5 — Promedio y comparación de temperaturas
Trabaje con el siguiente conjunto de datos:

Temperaturas registradas (en grados Celsius):
- Lunes: 18.5
- Martes: 21.0
- Miércoles: 19.8
Temperatura promedio de referencia de la semana: 20.0

```python
temperatura_lunes = 18.5
temperatura_martes = 21.0
temperatura_miercoles = 19.8
temperatura_promedio_semana = 20.0

# Calcular el promedio de los tres días

promedio = (temperatura_lunes + temperatura_martes + temperatura_miercoles) / 3
print(promedio)

# Calcular la diferencia entre martes y lunes

diferencia = temperatura_martes - temperatura_lunes
print(diferencia)

# Evaluar si el martes fue más caluroso que el promedio de referencia

print("¿El martes fue más caluroso que el promedio?",
      temperatura_martes > temperatura_promedio_semana)

# Evaluar si el miércoles fue igual al promedio de referencia

print("¿El miércoles tuvo la misma temperatura que el promedio?",
      temperatura_miercoles == temperatura_promedio_semana)

# Evaluar si el lunes fue el día más frío

print("¿El lunes fue más frío que el promedio?",
      temperatura_lunes < temperatura_martes and temperatura_lunes < temperatura_miercoles)
```

Ejercicio 10 — Intercambiar dos precios
Una ferretería tiene dos repuestos con precios mal registrados: el precio_tornillo quedó guardado como el de la tuerca, y viceversa. Corrija el error intercambiando los valores de las dos variables, sin asignar los valores correctos a mano — debe usar una variable temporal para el intercambio.

```python
precio_tornillo = 250
precio_tuerca = 180

# Valores antes de intercambiar

print("Antes de intercambiar:")
print("Precio tornillo:", precio_tornillo)
print("Precio tuerca:", precio_tuerca)

# Intercambiar usando una variable temporal

temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

# Valores después de intercambiar

print("Después de intercambiar:")
print("Precio tornillo:", precio_tornillo)
print("Precio tuerca:", precio_tuerca)
```
